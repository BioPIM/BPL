```python
import pandas as pd
import csv
```

```python
stats = [];
for idx,line in enumerate(open("traces.txt")):
    info = [x for x in line.replace(","," ").replace("[","").replace("]","").split(" ") if x!="" and "-" not in x ] 
    if len(info)<3: continue;
    info = [float(info[i]) for i in [3,4,5,7,9,11,13,16,17,19]]
    stats.append (info)
```

```python
df = pd.DataFrame(stats, columns=["rank","dpu","tasklet", "nbitemsPerPu", "multicore(32)", "upmem:all", "ratio", "mram_call", "mram_size", "upmem:exec"])
list(df.columns)
```

```python
def show(df,nbitemsPerPu):
    dfp = df[df["nbitemsPerPu"]==nbitemsPerPu];
    dfp.plot(x="tasklet",y=["multicore(32)","upmem:all","upmem:exec"], ylabel="time (s)", title="Selection Sort (nbitems/tasklets={})".format(nbitemsPerPu));
```

```python
for i in [6,10,12]:  show(df,1<<i)
```
