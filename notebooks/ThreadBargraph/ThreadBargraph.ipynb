```python
import pandas as pd

df = pd.read_csv("traces/bargraph.txt", sep='\\s+', header=None, names=["thread","task","begin","end","duration"])
groups = df.groupby("thread")

def bargraph(i,u,v,M,nbchar=150):
    i0=int(nbchar*float(u)/float(M))
    i1=int(nbchar*float(v)/float(M))
    res = str(i%10)*(i1-i0)
    return res 

M = df["end"].max()
b = groups.agg({"begin": lambda x: list(x)}).values.tolist()
d = groups.agg({"end"  : lambda x: list(x)}).values.tolist()
for i,(x,y) in enumerate(zip(b,d)): 
    l = "".join (bargraph(i,u,v,M) for i,(u,v) in enumerate(zip (x[0],y[0])))
    print ("{:2} {}".format(i,l))
```
