# Sketches comparison benchmark

```python
import pandas as pd
```

## Reading the benchmark results
The results are given by the benchmark binary as console output. We format a little bit this output in order to have a pandas dataframe.

```python
types={ 
    "native_multicore": "1.MC std", 
    "bpl_multicore"   : "2.MC BPL", 
    "sdk_pim"         : "3.PIM SDK", 
    "bpl_pim"         : "4.PIM BPL"};

header=[]
rows=[]
with open("results8.txt") as f:
    for l in [x.replace("\n","").split() for x in f.readlines()]:
        if len(header)==0:
            header=[l[0],l[2],l[4],l[6],l[8],l[10]]
        if l[1] in types:  l[1] = "{}".format(types[l[1]])
        rows.append([l[1],l[3],l[5],l[7],l[9],l[11]])

header = [x.replace(":","") for x in header]
header = [x.replace("#","") for x in header]

df = pd.DataFrame(rows, columns=header)
for c in ["ref","qry","nbunits","duration"]:
    df[c] = pd.to_numeric(df[c])
df
```

## Checking results
We check that, for the same configuration, the different runs provide the same result

```python
dfcheck = df.groupby(["ref","qry","nbunits"])["checksum"]
check = list(set((dfcheck.nunique()==1).tolist()))
print ("status: ", len(check)==1 and check[0]==True)
```

## Output table generation
We generate the result as a 

```python
df1 = df[['type', 'ref', 'qry', 'nbunits', 'duration']]

df2=df1.pivot(index=["ref","qry","nbunits"], columns=["type"], values=["duration"]) 
df2
```

```python
df2.to_latex(
    buf = "chapters/mash/table.tex",
    label = "tab:mash-results",
    bold_rows={"name": str.upper}, 
    float_format="{:.2f}".format, 
    column_format="lll|rr|rr",
    caption="Results of the sketches comparison benchmark")
```
