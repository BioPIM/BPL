```python
def display_speeduparch(dftask):
    tmp1=dftask[dftask["arch"]=="multicore"]
    tmp2=dftask[dftask["arch"]=="upmem"]
    tmp3=tmp1.merge(tmp2, on=["unit/nbcomp"])
    tmp3["speedup"] = tmp3["time_x"] / tmp3["time_y"]
    fig = px.line(tmp3, x="unit/nbcomp", y="speedup")
    fig.update_layout(title="Speedup upmem/multicore", yaxis_title="Speedup := T(multicore) / T(upmem)")
    fig.update_layout(title="Speedup muticore vs. upmem")
    fig.update_layout(hovermode="x unified")
    fig.update_layout(width=600,height=600)
    fig.update_traces(hovertemplate="%{y:4.1f}x",mode='lines+markers')
    fig.update_xaxes(title_text="ranks/threads")
    fig.show()
    return fig
```

```python
display_speeduparch(df).write_image("pang_compare_arch.png")
```

```python
def compute_speedup(df):
    dfTmp = pd.DataFrame()
    for arch in ["upmem","multicore"]:
        tmp = df[["arch","unit/nbcomp","time"]] [(df["arch"]==arch)].copy()
        if len(tmp)==0: continue;
        tmp["speedup"] = tmp[tmp["unit/nbcomp"]==1]["time"].iloc[0] / tmp["time"]
        dfTmp = tmp if len(dfTmp)==0 else pd.concat ([dfTmp,tmp])
    return df.merge (dfTmp, on=["arch", "unit/nbcomp","time"], how="left")

dftask = compute_speedup(df);

fig = px.line(dftask, x="unit/nbcomp", y="speedup", color="arch")
fig.update_layout(width=1*600,height=600, title="Speedup", yaxis_title="Speedup T(0)/T(x)")
fig.update_layout(title="Scalability per architecture")
fig.update_layout(hovermode="x unified")
fig.update_traces(hovertemplate="%{y:4.1f}x",mode='lines+markers')
fig.update_xaxes(title_text="ranks/threads")
fig.update_layout(legend=dict(x=0.05,y=.75))
fig.show()
fig.write_image("pang_scalability.png")
```
